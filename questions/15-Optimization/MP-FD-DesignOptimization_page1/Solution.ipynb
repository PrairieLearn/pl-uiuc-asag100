```python
nelx = 3
nely = 2
xvec = [np.random.choice([0.0,0.2,0.3,0.5,0.7,1.0]) for i in range(nelx*nely)]
xvec = np.round(xvec,2)
X = xvec.reshape((nelx,nely))

plt.imshow(-X.T, cmap='gray',interpolation='none' )

for i in range(nelx):
    for j in range(nely):
        plt.text(i,j,X[i,j], color='orange',fontsize=18,weight='bold')
plt.show
```
