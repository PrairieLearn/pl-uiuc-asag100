 Write a function <code>MaxSumIncSub</code> that given an input list <code>L</code>, computes the maximum sum of any increasing subsequence in <code>L</code>. Recall that a subsequence is any subset of the elements of <code>L</code> considered in the same order in which they occurred in <code>L</code>. The function <code>MaxSumIncSub</code> should therefore compute the maximum sum that any increasing subsequence in <code>L</code> generates. Your function should run in $O(n^2)$ time.

Consider the following examples.

* <code>MaxSumIncSub([])</code> should return <code>None</code>, since there is no increasing subsequence that can be picked.
* <code>MaxSumIncSub([2])</code> should return <code>2</code>, since there is only one increasing subsequence in this case - [2].
* <code>MaxSumIncSub([1,2])</code> should return <code>3</code>. In this case, the following are all increasing subsequences - [1], [2], [1,2]. The maximum sum any increasing subsequence has in this case is 3, corresponding to the increasing subsequence [1,2].
* <code>MaxSumIncSub([2,1])</code> should return <code>2</code>. In this case, the following are all increasing subsequences - [2], [1]. Note that [2,1] is NOT an increasing subsequence. The maximum sum any increasing subsequence has in this case is 2, corresponding to the increasing subsequence [2].
* <code>MaxSumIncSub([7, 2, 1, 3])</code> should return <code>7</code>. In this case, the following are all increasing subsequences - [7], [2], [1],[3], [2,3], [1,3]. The maximum sum any increasing subsequence has in this case is 7, corresponding to the increasing subsequence [7].
* <code>MaxSumIncSub([5, 2, 2, 4])</code> should return <code>6</code>.  In this case, the following are all increasing subsequences - [5], [2], [4], [2,4]. The maximum sum any increasing subsequence has in this case is 6, corresponding to the increasing subsequence [2,4].
* <code>MaxSumIncSub([5,-2,1,2,0,13,4])</code> should return <code>18</code>. In this case there are too many increasing subsequences to list them out. The maximum sum of any increasing subsequence in this case is 18 - corresponding to the subsequence [5,13].

This problem can be solved using Dynamic programming. As always, there are 3 steps we need to follow before coding the solution.
1. Identify the sub-problems
2. Write a recursive solution to calculate the result of the sub-problems.
3. Identify which among the subproblems is the answer we are looking for.

For this problem, we can approach the solution as follows.
1. *Sub-problems:* Define $S(i)$ to be the maximum sum of any increasing subsequence ending at position $i$. Such an increasing subsequence must include $L[i]$.

 For example, when $L = [7, 2, 1, 3]$, and $i = 2$, then $S(2)$ will be $1$, since the maximum sum of any increasing subsequence at position $2$ is $1$ - corresponding to the increasing subsequence $[1]$.

2. *Recursive Solution:* For any increasing subsequence that ends at position $i$, the possible increasing subsequences that $L[i]$ can be added to must end with some $L[j]$ satisfying $L[j] < L[i]$. Among all such choices, we must pick the best one - that is a value of $j < i$ satisfying $L[j] < L[i]$ that maximizes $S(j) + L[i]$. Thus, we can say
$$S(i) = \max_{j<i : L[j] < L[i]} S(j) + L[i].$$
In addition, $S(0) = L[0]$.

3. *Answer:* The answer we are looking for is $\max_{0 \leq i \leq n-1}S(i)$. This is because the maximum sum can correspond to an increasing subsequence that can end at any of the $n$ positions - and we have to pick the best of the lot.
